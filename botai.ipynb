```python
import nest_asyncio
nest_asyncio.apply()

import discord
from discord.ext import commands

import tf_keras as keras
from tf_keras.models import load_model
from PIL import Image, ImageOps
import numpy as np

intents = discord.Intents.default()
intents.message_content = True

bot = commands.Bot(command_prefix='!', intents=intents)

def get_class(model, labels, image):

  np.set_printoptions(suppress=True)
  model = load_model(model, compile=False)

  class_names = open(labels, "r").readlines()
  data = np.ndarray(shape=(1, 224, 224, 3), dtype=np.float32)
  image = Image.open(image).convert("RGB")

  size = (224, 224)
  image = ImageOps.fit(image, size, Image.Resampling.LANCZOS)
  image_array = np.asarray(image)
  normalized_image_array = (image_array.astype(np.float32) / 127.5) - 1
  data[0] = normalized_image_array


  prediction = model.predict(data)
  index = np.argmax(prediction)
  class_name = class_names[index]
  confidence_score = prediction[0][index]

  clase_limpia = class_name.strip()
  return f"La imagen es un {clase_limpia} con una confianza de {confidence_score * 100}%"

@bot.event
async def on_ready():
    print(f'Bot iniciado{bot.user}')

@bot.command()
async def hello(ctx):
  await ctx.send("Holaaaaaa")

@bot.command()
async def detectar(ctx):
  if ctx.message.attachments:
    for attachment in ctx.message.attachments:
      await attachment.save(attachment.filename)
      await ctx.send(get_class("keras_model.h5", "labels.txt", attachment.filename))
  else:
    await ctx.send("No agregaste ninguna imagen :c")


bot.run("MTQ4NjUxMjc3Mjc0MDI4NDQ5Ng.GtQ_nj.9XnDUzbZ3XHA97eVR1F_g2IkvBUwPvjYjtvAVA")
```
